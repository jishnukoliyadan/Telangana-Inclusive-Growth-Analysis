```python
import os
import wget
import requests
from time import sleep
```

```python
def create_URL(api: str) -> str:
    URL = f"https://data.telangana.gov.in/api/1/metastore/schemas/dataset/items/{api}?show-reference-ids=true"
    print(URL)
    return URL

def get_urls(url: str) -> str:
    req = requests.get(url)
    urls = [item.get('data').get('downloadURL') for item in req.json().get('distribution')]
    for url in urls[:5]:
        print(url)
    print()
    urls = [url.replace('%20', ' ') for url in urls if url.replace('%20', ' ') not in unwanted_datasets]
    return urls

def download_files(urls: str, folder: str) -> None:
    os.makedirs(folder, exist_ok = True)
    # https://stackoverflow.com/a/45978493
    for item in urls:
        filename = item.split('/')[-1]
        try: wget.download(item, out = folder)
        except: print(item)
        finally: sleep(0.25)

def get_data(api: str, folder: str) -> None:
    URL = create_URL(api)
    urls = get_urls(URL)
    download_files(urls, folder)
```

```python
unwanted_datasets = ['https://data.telangana.gov.in/sites/default/files/OnlineVehicleSaleupdated_0.csv',
                     # Transport Vehicle Online Sales Data 2014 to 24-10-2018
                     'https://data.telangana.gov.in/sites/default/files/ts_transport_vehicle_online_sales_25_10_2018to31_12_2018.csv',
                     # Transport Vehicle Online Sales Data 25-10-2018 to 31-12-2018
                    ]
```
